## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from sklearn.neural_network import MLPClassifier
```

## Simple Classification Demo (one linearly seperable set)

```python
df=pd.read_csv('data/linear_blobs_small.csv')
X=df[['x1','x2']]
y=df['label']
```

```python
clf=Perceptron()
clf.fit(X,y)

# Accuracy
df['pred']=clf.predict(X)

accuracy_score(y,df['pred'])
```

```python
# Quick plot
plt.scatter(df['x1'],df['x2'],c=df['pred'])
```

2. When a linear model struggles (non-linear rings)

```python
df=pd.read_csv('data/rings_small.csv')
X= df[['x1','x2']].values
y=df['label'].values
```

```python
clf=Perceptron()
clf.fit(X,y)

df['pred']=clf.predict(X)

accuracy_score(y,df['pred'])
```

```python
plt.scatter(df['x1'],df['x2'],c=df['pred'])
```

3. Tiny Neural Net

```python
# Small MLP (hidden layers give non-linearity)

mlp=MLPClassifier(hidden_layer_sizes=(8,), activation='relu')
mlp.fit(X,y)
```

```python
df['pred']=mlp.predict(X)

accuracy_score(y,df['pred'])
```
