```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
```

### 1. Create a linearly seperable-ish dataset+scatter

```python
np.random.seed(0)
```

```python
# Synthetic Linear-ish blobs
n = 250
X_lin = np.random.normal(0, 1, size=(n, 2))
y_lin = (X_lin[:, 0] + 0.8 * X_lin[:, 1] > 0.2).astype(int)
X_lin = X_lin + np.random.normal(0, 0.25, size=X_lin.shape) # small noise
X_lin
```

```python
# plot linear data
plt.figure()
plt.scatter(X_lin[:, 0], X_lin[:, 1], c=y_lin, cmap='bwr', alpha=0.7)
plt.title('Linearly Separable Data')
plt.xlabel('x1')
plt.ylabel('x2')
```

### 2. Fit a Perceptron on the linear dataset

```python
# Perceptron on linear data

perc_lin=Perceptron()
perc_lin.fit(X_lin,y_lin)

pred_linear=perc_lin.predict(X_lin)
```

```python
accuracy_score(y_lin,pred_linear)
```
