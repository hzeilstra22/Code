```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron
from sklearn.neural_network import MLPClassifier
```

### 1. Perceptron on linearly seperable data

```python
np.random.seed(0)
X = np.random.randn(200, 2)
y = (X[:,0] + X[:,1] > 0).astype(int)
```

```python
X
```

```python
plt.scatter(X[y==0,0],X[y==0,1],label="Class 0")
plt.scatter(X[y==1,0],X[y==1,1],label="Class 1")
plt.title('Linearly Seperable Data')
plt.legend()
plt.show()
```

```python
# Train Perceptron
perc=Perceptron()
perc.fit(X,y)
pred=perc.predict(X)

print("Weights:", perc.coef_)
print('Bias:', perc.intercept_)
print('Training Accuracy:', perc.score(X,y))
```

### 2. Perceptron on XOR

```python
# XOR dataset
X_xor=np.array([[0,0], [0,1], [1,0], [1,1]])
y_xor=np.array([0,1,1,0])
```

```python
X_xor
```

```python
plt.scatter(X_xor[y_xor==0,0],X_xor[y_xor==0,1],label="Class 0")
plt.scatter(X_xor[y_xor==1,0],X_xor[y_xor==1,1],label="Class 1")
plt.title('XOR Problem (Not Linearly Sepable)')
plt.legend()
plt.show()
```

```python
# Look at what perceptron actually learned on XOR
perc2=Perceptron()
perc2.fit(X_xor,y_xor)

print("Weights:", perc2.coef_)
print('Bias:', perc2.intercept_)
print('Training Accuracy:', perc2.score(X_xor,y_xor))
```

```python
pred=perc2.predict(X_xor)
print('predictions:', pred)
print('True labels:', y_xor)
```

### 3. MLP on XOR

```python
mlp=MLPClassifier(hidden_layer_sizes=[4,], activation='relu', random_state=0,max_iter=1000)

mlp.fit(X_xor, y_xor)

mlp.score(X_xor,y_xor)
```

```python
preds=mlp.predict(X_xor)
print('preds:', preds)
print("True Labels:", y_xor)
```
