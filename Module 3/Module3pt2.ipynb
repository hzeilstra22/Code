### 1. The forward pass

```python
import numpy as np

X=np.array([[0,0],[0,1],[1,0],[1,1]],dtype=float)
y=np.array([0,1,1,0])

def sigmoid(z):
    return 1/(1+np.exp(-z))

def forward(X, w1, b1, w2, b2):
    h=np.maximum(0,X @ w1 + b1)
    return (sigmoid(h @ w2 + b2).ravel())

def accuracy(probs, y):
    return((probs >= 0.5).astype(int)==y).mean()

rng=np.random.default_rng(0)
w1, b1=rng.normal(0,1,(2,2)), rng.normal(0,1,2)
w2, b2=rng.normal(0,1,(2,1)), rng.normal(0,1,1)

probs= forward(X, w1, b1,w2,b2)
print("probs:", probs.round(3))
print("pred:", (probs>=0.5).astype(int))
print("target:", y)
print("accuracy:", accuracy(probs, y))
```

### 2. Lots of random weights

```python
best, best_probs=0, None
rng=np.random.default_rng(0)
num_tries=10000

for i in range(num_tries):
    w1, b1=rng.normal(0,1,(2,2)), rng.normal(0,1,2)
    w2, b2=rng.normal(0,1,(2,1)), rng.normal(0,1,1)
    a=accuracy(forward(X, w1,b1,w2,b2),y)
    if a>best:
        best=a
        best_probs=forward(X,w1,b1,w2,b2)
print(f'Best after {num_tries} tries: {best}')
print(f'predictions: {(best_probs>=0.5).astype(int)} target: {y}')
```
