### Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import random
```

### Binary Step

```python
def binary_step(x):
# Binary Step Activation function
# Input: A number
# Output: 1 or 0 depending on if the input is greater than or less than the threshold, using 0 here
    if x < 0:
        return 0
    else: 
        return 1
```

```python
# plot binary step function on interval [-10,10]
df=pd.DataFrame({}) # initialize dataframe for plotting
df['x']=np.linspace(-10,10,100) # generate 100 rows of data between -10 and 10
df['binary_step_y']=df['x'].apply(binary_step) # apply activation function on generated data

plt.plot(df['x'],df['binary_step_y'])
plt.title('Binary Step Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/binary_step.png')
plt.show()
```

### Linear

```python
def linear(x):
# Linear Activation function
# Input: A number
# Output: The same number
    return x
```

```python
df['linear_y']=df['x'].apply(linear)

plt.plot(df['x'],df['linear_y'])
plt.title('Linear Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/linear.png')
plt.show()
```

### Sigmoid

```python
def sigmoid(x):
# Sigmoid Activation function
# Input: A number
# Output: A number between 0 and 1 according to formula f(x)=1/(1+e^-x)
    return 1/(1+np.exp(-x))
```

```python
# plot sigmoid function on interval [-10,10]
df['sigmoid_y']=df['x'].apply(sigmoid)

plt.plot(df['x'],df['sigmoid_y'])
plt.title('Sigmoid Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/sigmoid.png')
plt.show()
```

### Tanh

```python
def tanh(x):
    # Tanh Activation function
    # Input: A number
    # Output: A number between -1 and 1 according to formula f(x)=(e^x-e^-x)/(e^x+e^-x)
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))
```

```python
# plot tanh function on interval [-10,10]
df['tanh_y']=df['x'].apply(tanh)

plt.plot(df['x'],df['tanh_y'])
plt.title('Tanh Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/tanh.png')
plt.show()
```

### ReLU

```python
def relu(x):
# relu Activation function
# Input: A number
# Output: x if x>0 otherwise output is 0
    if x>0:
        return x
    else:
        return 0
```

```python
# plot ReLU function on interval [-10,10]
df['relu_y']=df['x'].apply(relu)

plt.plot(df['x'],df['relu_y'])
plt.title('ReLU Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/relu.png')
plt.show()
```

### Leaky ReLU

```python
def leaky_relu(x):
# leaky relu Activation function
# Input: A number
# Output: x if x>=0 otherwise x * alpha where alpha is a small constant; set to 0.01 here
    if x>=0:
        return x
    else:
        return 0.01*x
```

```python
# plot Leaky ReLU function on interval [-10,10]
df['leaky_relu_y']=df['x'].apply(leaky_relu)

plt.plot(df['x'],df['leaky_relu_y'])
plt.title('Leaky ReLU Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/leaky_relu.png')
plt.show()
```

### Softmax

```python
def softmax(x, array):
# softmax Activation function
# Input: A vector of numbers
# Output: Probability distribution, where each value is likelihood of that class. 
    # Each output lies between 0 and 1 and the sum of all output values is 1
    exp_x=np.exp(x)
    exp_array=np.exp(array)
    return exp_x / np.sum(exp_array)
```

```python
# plot softmax function on interval [-10,10]
df['softmax_y']=df['x'].apply(softmax,array=df['x'])

plt.plot(df['x'],df['softmax_y'])
plt.title('Softmax Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.savefig('images/softmax.png')
plt.show()
```
